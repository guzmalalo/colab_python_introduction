# Le langage Python 

Python est un langage **interprété**, chaque instruction est analysée, traduite et exécutée au moment du lancement. Aucun exécutable (`.exe`) n'est créé et il n'y a pas de fonction `main()`.

**En Python, les instructions s'écrivent directement dans le fichier source :**


````python
i = 0
print('Hello world!')
print('i vaut :', i)
````

# Du C au Python...

- Pas d'accolades **`{}`**, c'est l'**indentation** et **`:`** qui délimitent les **blocs d'instructions**.
- Pas de point-virgule **`;`**, **une instruction par ligne**.
- Les parenthèses **`()`** servent notamment aux **appels de fonctions** et à **grouper des expressions** : `(2 + 3) * 4`.


````python
i = 0

while i <= 100:
    i = (i + 2) * 3
    print('i = ', i)

print('Valeur finale de i :', i)  # Notez l'absence de %d
````

**Note :** Utilisez **4 espaces par niveau d'indentation**, sans tabulations.


# Bloc d'instructions

- Les instructions d'un même bloc ont **la même indentation**.
- Un bloc peut contenir **un autre bloc**, indenté d'un niveau supplémentaire.
- Revenir au niveau précédent marque **la fin du bloc**.


````python
i = 0
while i < 3:
    print('i =', i)          # Bloc du while
    if i == 2:
        print('Dernier tour')  # Bloc du if
    i = i + 1              # Retour au bloc du while

print('Fin')               # En dehors des deux blocs
````

# Commentaires

Tout ce qui se trouve à droite d'un **croisillon** (**`#`**) est interprété comme un **commentaire**.


````python
# Cette ligne est un commentaire
i = 0

while i <= 5: # Cette partie en gris, aussi
    # Ceci est un commentaire sur
    # deux lignes, dans un bloc
    i = i + 1
    print('i = ', i)

# Ce print n'est pas dans le bloc de la boucle while
print('Valeur finale de i : ', i)
````

***Note :** Chaque commentaire doit être indenté au même niveau que son bloc.*

# Fonctions

## À quoi sert une fonction ?

Une **fonction** regroupe des instructions pour réaliser une tâche précise.

- **Nommer un calcul** pour rendre le programme plus lisible.
- **Réutiliser le même calcul** avec différentes données.
- **Éviter de recopier** les mêmes instructions.

Pour calculer l'aire d'un rectangle : **$A = L \times l$**.

**Entrées :** longueur et largeur → **Fonction :** `aire_rectangle` → **Résultat :** aire en cm².


# Fonctions

## Nom, entrées et retour


````python
def aire_rectangle(longueur, largeur):
    aire = longueur * largeur
    return aire
````

- **Nom** : `aire_rectangle` décrit la tâche.
- **Entrées** : `longueur` et `largeur` sont les **paramètres**, entre parenthèses.
- **Corps** : les instructions indentées effectuent le calcul.
- **Retour** : **`return`** termine la fonction et renvoie le résultat.

**Note :** **`def`** **déf**init la fonction ; l'**appel** déclenche le calcul.


# Fonctions

## Appeler une fonction et utiliser son résultat

On écrit le **nom**, puis les **valeurs d'entrée** entre parenthèses, dans le même ordre que les paramètres.


````python
surface = aire_rectangle(3, 2)  # Longueur : 3 cm, largeur : 2 cm
print('Aire :', surface, 'cm²')

surface = aire_rectangle(4, 2)  # Même fonction, autre longueur
print('Aire :', surface, 'cm²')
````

**Note :** **`return`** fournit un résultat utilisable dans le programme ; **`print()`** affiche une information dans la console.


# Fonctions

## Retourner deux résultats

Une fonction peut retourner plusieurs résultats, séparés par des virgules. Lors de l'appel, on utilise autant de variables que de résultats retournés : `x, y` pour deux résultats, `x, y, z` pour trois, etc.

Par exemple, pour un rectangle : **$A = L \times l$** et **$P = 2(L + l)$**.


````python
def aire_et_perimetre(longueur, largeur):
    aire = longueur * largeur
    perimetre = 2 * (longueur + largeur)
    return aire, perimetre
x, y = aire_et_perimetre(3, 2)
print('Aire :', x, 'cm²')
print('Périmètre :', y, 'cm')
````

**Note :** **`x`** reçoit l'aire et **`y`** le périmètre, dans **l'ordre du `return`**.
