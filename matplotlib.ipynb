# Définitions

Matplotlib affiche vos données sur des **Figures** dont chacune peut contenir un ou plusieurs **Axes**.

Une **Figure** est le conteneur principal pour nos **Axes**, cela peut être une fenêtre, une zone dans un notebook, etc.

Un **Axe** est la zone sur laquelle sont dessinés les différents points en fonction de leurs coordonnées (**a.k.a. graphique**).

**Note :** Une Figure est comme un `display` dans Allegro.


# Création d'une figure

La méthode la plus simple pour créer une **Figure** avec un **Axe** est d'utiliser **`plt.subplots()`**.


````python
import matplotlib.pyplot as plt

fig, ax = plt.subplots()

display(fig)
plt.close(fig)
````

# Création d'une figure

On peut utiliser la méthode **`plot()`**, de notre Axe `ax`, pour commencer à dessiner nos données sous la forme d'un **graphique x-y 2D** :

**Syntaxe :**

```shell
ax.plot(x, y, fmt, x2, y2, fmt, ...)
```

**Exemple :**


````python
fig, ax = plt.subplots()
#            x              y
ax.plot([1, 2, 3, 4], [1, 4, 2, 3])

display(fig)
plt.close(fig)
````

**Note :** `fmt` est le format de la courbe.


# Propriétés

L'Axe `ax` est un objet contenant toutes les propriétés et les méthodes pour manipuler l’apparence de vos graphiques :

- Titre, légendes
- Échelle x et y
- Couleur
- Marqueurs, type de ligne
- Et plus  [API Axes](https://matplotlib.org/stable/api/axes_api.html)

**Note :** objet ≈structure en **C** + méthodes


# Propriétés

Modifions, par exemple, le titre du graphique, le nom des axes et les limites (x et y).

- Titre


````python
ax.set_title("Hello Python")

display(fig)
plt.close(fig)
````

- Axes


````python
ax.set_xlabel("x")
ax.set_ylabel("y")

display(fig)
plt.close(fig)
````

- Limites


````python
ax.axis([0, 5, 0, 5])
#   [xmin, xmax, ymin, ymax]

display(fig)
plt.close(fig)
````

# Propriétés

Un graphique (tracé) contient plusieurs propriétés en fonction de son [type](https://matplotlib.org/stable/api/axes_api.html#plotting).

Pour un graphique **2D (x, y)**, on peut définir ces propriétés au moment du `plot()`.

**Propriétés courantes**

- `marker` = Marqueur
- `linestyle` = Type
- `color` = Couleur de ligne
- `label` = Légende (nom du graphe)
- `linewidth` = Épaisseur


# Propriétés

**Marqueurs :**


````python
x = [1, 2, 3, 4]
y = [1, 4, 2, 3]
fig, ax = plt.subplots()
ax.plot(x,y, marker = "s")

display(fig)
plt.close(fig)
````

**Quelques marqueurs :**

- `"."` : Point
- `"o"` : Cercle
- `"^"` : Triangle
- `"s"` : Carré

 [Liste complète](https://matplotlib.org/stable/api/markers_api.html#module-matplotlib.markers)


# Propriétés

**Couleur :**


````python
x = [1, 2, 3, 4]
y = [1, 4, 2, 3]
fig, ax = plt.subplots()
ax.plot(x,y, color = "r")

display(fig)
plt.close(fig)
````

**Quelques couleurs :**

- `"r"` : Rouge ou `(1,0,0)` (RGB)
- `"g"` : Vert ou `"#00ff00"` (HEX)
- `"b"` : Bleu
- `"k"` : Noir

 [Liste complète](https://matplotlib.org/3.5.0/tutorials/colors/colors.html)


# Propriétés

**Type de ligne** :


````python
x = [1, 2, 3, 4]
y = [1, 4, 2, 3]
fig, ax = plt.subplots()
ax.plot(x,y, linestyle = ":")

display(fig)
plt.close(fig)
````

**Quelques styles** :

- `"-"` : Solide ou `"solid"`
- `":"` : Pointillé ou `"dotted"`
- `"--"` : Tirets ou `"dashed"`
- `"-."` : Tiret-point ou `"dashdot"`

 [Liste complète](https://matplotlib.org/stable/gallery/lines_bars_and_markers/linestyles.html)


# Propriétés

**Exemple complet** :


````python
x = [1, 2, 3, 4]
y = [1, 4, 2, 3]
fig, ax = plt.subplots()
ax.plot(x,y, marker="s",
             color="r",
             linestyle = ":")

display(fig)
plt.close(fig)
````

C'est long , heureusement on peut utiliser un format plus court (`fmt`).

```bash
fmt = '[marker][color][linestyle]'
# exemple
fmt = 'sr:'
```


# Propriétés

**Exemple complet** :


````python
x = [1, 2, 3, 4]
y = [1, 4, 2, 3]
fig, ax = plt.subplots()
ax.plot(x,y,'sr:')

display(fig)
plt.close(fig)
````

Pour les autres propriétés, comme la légende (`label`), nous sommes obligés de définir la propriété.


````python
ax.plot(x, y, 'sr:', label="Data 1")
ax.legend()

display(fig)
plt.close(fig)
````

**Note :** `ax.legend()` rend les légendes visibles.


# Autres propriétés

- Taille de la figure

```
fig, ax = plt.subplots(figsize=(5, 2.7))
```

- Grille


````python
ax.grid(True) # + format (optionnel)
# color=(0.75,0.75,0.75),
# linestyle='-', linewidth=2)

display(fig)
plt.close(fig)
````

- Échelle


````python
ax.set_yscale("log")
ax.set_xscale("linear") # Par défaut

display(fig)
plt.close(fig)
````

# Multi-graphes

Pour ajouter plusieurs séries de données sur un **même graphique** (Axe). Il suffit d'ajouter les séries :

- Au même moment :


````python
x  = [1, 2, 3, 4]
y1 = [1, 4, 2, 3]
y2 = [2, 1, 5, 6]
fig, ax = plt.subplots()
ax.plot(x, y1, 'sr:', x, y2, 'ok-')

display(fig)
plt.close(fig)
````

:-1: Mais la gestion des légendes et des propriétés de chaque courbe devient difficile.


# Multi-graphes

Pour ajouter plusieurs séries de données sur un **même graphique** (Axe). Il suffit d'ajouter les séries :

- Une par une :


````python
x  = [1, 2, 3, 4]
y1 = [1, 4, 2, 3]
y2 = [2, 1, 5, 6]
fig, ax = plt.subplots()
ax.plot(x, y1, 'sr:', label="Data 1")
ax.plot(x, y2, 'ok-', label="Data 2")
ax.legend()

display(fig)
plt.close(fig)
````

# Multi-graphes

Pour ajouter plusieurs séries de données sur des **graphes différents** (Axes), mais sur la **même figure**, on utilise la version généralisée de `plt.subplots()`.


````python
# nrows : nombre d'axes (vertical)
# ncols : nombre d'axes (horizontal)
# nrows=ncols=1 par défaut.
nrows, ncols = 2, 2
fig, axs = plt.subplots(nrows, ncols)

display(fig)
plt.close(fig)
````

**Exemple :**


````python
fig, axs = plt.subplots(3,3)

display(fig)
plt.close(fig)
````

**Note :** `axs` est maintenant un tableau d'axes.


# Multi-graphes

Pour ajouter plusieurs séries de données sur des **graphes différents** (Axes), mais sur la **même figure**, on utilise la version généralisée de `plt.subplots()`.

**Un autre exemple :**


````python
x  = [1, 2, 3, 4]
y1 = [1, 4, 2, 3]
y2 = [2, 1, 5, 6]
fig, axs = plt.subplots(1,2)
axs[0].plot(x, y1, 'sr:', label="Data 1")
axs[1].plot(x, y2, 'ok-', label="Data 2")
for ax in axs:
  ax.legend()

display(fig)
plt.close(fig)
````

# Matplotlib & NumPy

Vous pouvez utiliser les tableaux **NumPy** avec les méthodes fournies par Matplotlib. C'est logique et fortement recommandé.

- **NumPy** : manipule les données
- **Matplotlib** : affiche le résultat

**Exemple**


````python
import numpy as np

x = np.arange(0,2.0*np.pi, 0.25)
fig, ax = plt.subplots()
ax.plot(x,np.sin(x),'xr-', label='sin')
ax.plot(x,np.cos(x),'ob-', label='cos')
y = np.cos(x)+np.sin(x)
ax.plot(x,y,'^g:',label='sin+cos')
ax.legend()

display(fig)
plt.close(fig)
````

# Un exemple complet


````python
# Importation bibliothèques
import matplotlib as mpl
import matplotlib.pyplot as plt
import numpy as np
# Données
x = np.arange(0,2.0*np.pi, 0.25)
# Figure et axes
fig, ax = plt.subplots()
ax.set_title('Sinus and Cosinus')
ax.set_xlabel('x')
ax.set_ylabel('y')
# Plots
ax.plot(x,np.sin(x),'xr-', label='sin')
ax.plot(x,np.cos(x),'ob-', label='cos')
y = np.cos(x)+np.sin(x)
ax.plot(x,y,'^g:',label='sin+cos')
ax.grid(True)
ax.legend()

display(fig)
plt.close(fig)
````
